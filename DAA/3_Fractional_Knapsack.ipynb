```python

def fractional_knapsack(values, weights, capacity):
    n = len(values)
    items = [(values[i], weights[i], values[i] / weights[i], i) for i in range(n)]
    items.sort(key=lambda x: x[2], reverse=True)

    remaining = capacity
    total_value = 0.0
    fraction = [0.0] * n

    for value, weight, ratio, index in items:
        if remaining <= 0:
            break

        if weight <= remaining:
            fraction[index] = 1.0
            total_value += value
            remaining -= weight
        else:
            fraction[index] = remaining / weight
            total_value += value * fraction[index]
            remaining = 0

    return total_value, fraction


n = int(input("Enter number of items: "))

print("Enter values:")
values = list(map(float, input().split()))

print("Enter weights:")
weights = list(map(float, input().split()))

capacity = float(input("Enter knapsack capacity: "))

if len(values) != n or len(weights) != n:
    print("Error: number of values and weights must equal the number of items.")

elif any(w <= 0 for w in weights) or capacity < 0:
    print("Error: weights must be positive and capacity cannot be negative.")

else:
    total_value, fraction = fractional_knapsack(values, weights, capacity)

    print("\nFractions of items selected:")
    for i in range(n):
        print(f"Item {i + 1}: {fraction[i]}")

    print(f"\nMaximum value = {total_value}")
```

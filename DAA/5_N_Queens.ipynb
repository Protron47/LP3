```python

def is_safe(row, col):
    for r in range(n):
        c = queens[r]
        if c != -1 and r != row:
            if c == col or abs(r - row) == abs(c - col):
                return False
    return True


def print_board():
    for r in range(n):
        print(" ".join("Q" if queens[r] == c else "." for c in range(n)))


def solve(row):
    global count

    if row == n:
        count += 1
        print("\nSolution", count)
        print_board()
        return

    if row == first_row:
        solve(row + 1)
        return

    for col in range(n):
        if is_safe(row, col):
            queens[row] = col
            solve(row + 1)
            queens[row] = -1


# Main program
n = int(input("Enter number of queens: "))
first_row, first_col = map(int, input("Enter first queen row and column: ").split())

if n < 1 or not (0 <= first_row < n and 0 <= first_col < n):
    print("Invalid input")
else:
    queens = [-1] * n
    queens[first_row] = first_col
    count = 0

    print("\nInitial Board:")
    print_board()

    solve(0)
    print("\nTotal solutions =", count)
#4e0 2
```

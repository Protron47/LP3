```python
import heapq
from collections import Counter

def huffman(text):
    freq = Counter(text)
    heap = [[f, [c, ""]] for c, f in freq.items()]
    heapq.heapify(heap)

    while len(heap) > 1:
        a = heapq.heappop(heap)
        b = heapq.heappop(heap)

        for x in a[1:]:
            x[1] = "0" + x[1]
        for x in b[1:]:
            x[1] = "1" + x[1]

        heapq.heappush(heap, [a[0] + b[0]] + a[1:] + b[1:])

    codes = dict(heap[0][1:])
    print("Huffman Codes:", codes)
    print("Encoded:", "".join(codes[c] for c in text))


text = input("Enter text: ")
huffman(text)
```
